### 2\. CREATE TABLE BOOKS

  

\* WE CAN CREATE THE TABLE BY DEFINING COLUMNS AND THEIR PROPERTIES

```sql
DROP TABLE IF EXISTS BOOKS;

CREATE TABLE BOOKS(
    TITLE VARCHAR(80) NOT NULL,
    AUTHOR VARCHAR(80) NOT NULL,
    COVER VARCHAR(256) DEFAULT('unknown.jpg'),
    PRICE DECIMAL DEFAULT(0),
    RATING DECIMAL NOT NULL,
    DESCRIPTION  VARCHAR(2000)  
);
```

## 4\. Query for the books:  SELECT  

- The most used query of all
- Has many forms and parameters
- Helps us fetch various details
- We will discuss most variants separately
- For now we will query and check out other functionalities

  

4.1 Select specific columns from books (\* means all columns)

```sql
SELECT
    TITLE,
    AUTHOR
FROM
    BOOKS
```

## UPDATE A RECORD

  

- We Can update the records by calling update command and set

```sql
UPDATE
    BOOKS
SET
    PRICE=0
```

## UPDATE CONDITIONALLY BY ADDING WHERE

```sql
UPDATE
    BOOKS
SET
    AUTHOR='Vivek'
WHERE
    AUTHOR='Vivek Dutta Mishra'
```

DELETE RECORDS

  

- Again we can delete the row based on a condition
- if no condition given the entire table is deleted
-

```sql
DELETE
    FROM BOOKS
WHERE
    AUTHOR='Vivek'
```

# Alter Table af table or column

- WE can 
    - Add column
    - Alter column
    - change constraints

### ALTER TO ADD A NEW COLUMN

- Make sure alter doesn't make the current row or invalid
    - Alter will fail
- If you are adding a new column make sure
    - new column is either allowing null or has default

```sql
ALTER TABLE BOOKS
    ADD TAGS VARCHAR(50) NOT NULL
;
```

```sql
ALTER TABLE BOOKS
    ADD TAGS VARCHAR(50) NOT NULL DEFAULT('')
;
```

### WE CAN DROP OR ALTER COLUMNS

```sql
ALTER TABLE BOOKS
    ALTER COLUMN PRICE FLOAT NOT NULL
    
```

```sql
DELETE FROM BOOKS
```

```sql
ALTER TABLE BOOKS
    DROP COLUMN DESCRIPTION
```

```sql
ALTER TABLE BOOKS
    DROP CONSTRAINT DF__BOOKS__PRICE__6383C8BA
    
```
