# 2\. FILTERING THE ROWS

- FILTERING ALLOWS US TO ACCESS A SUBSET OF ROWS OR DATA FROM OUR TABLE
- SOME FILTERING CLAUSES LIKE 'WHERE' WORKS WITH OTHER STATEMENTS LIKE UPDATE AND DELETE

## 2.1  WHERE CLAUSE

- <span style="color: var(--vscode-foreground);">WORKS MORE LIKE JS FILTER</span>
- WE DEFINE A MATCHER CLAUSE TO FILTER THE RECORDS

```sql
SELECT  TITLE, AUTHOR, PRICE
FROM    BOOKS
WHERE   PRICE < 200
```

### WHERE CHECKS FOR DATA IN CASE INSENSETIVE WAY

- WHERE AUTHOR NAME IS MENTIONED IN ALL LOWER CASE 
- STILL MATCHES TO NAMES MENTIONED IN PROPER CASE

```sql
SELECT TITLE, AUTHOR, PRICE
FROM BOOKS
WHERE AUTHOR = 'jeffrey archer'
```

### WHERE BY DEFAULT MATCHES WHOLE TEXT

```sql
SELECT TITLE, AUTHOR, PRICE FROM BOOKS
WHERE AUTHOR='Vivek'
```

### PARTIAL MATCH USING LIKE

- Like replaces = in comparision
- It is a partial match using the wildcard "%" to mean anything
- Generally used before or after the text match
    - Vivek%  ---\> starting with Vivek and then any number of character after it
    - %Vivek ----\> ending with the word Vivek
    - %Vivek% ---\> containing the word Vivek

```sql
SELECT TITLE, AUTHOR, PRICE 
FROM BOOKS
WHERE AUTHOR LIKE 'VIVEK%'
```

```sql
SELECT * FROM BOOKS
WHERE TITLE LIKE '%Prison%'
```

# Row limiting clauses

- We can request to return only a section of result rather than whole result
- We can do by applying clauses like
    - top
    - bottom

```sql
SELECT     
    TOP 3
    TITLE, AUTHOR, PRICE
FROM
    BOOKS
WHERE AUTHOR LIKE '%ROWLING%'
```

### ORDERING THE RESULT

- ORDER BY 
-

```sql
SELECT TITLE, AUTHOR
FROM BOOKS
WHERE AUTHOR NOT LIKE '%ROWLING%'
ORDER BY AUTHOR
```

### WE CAN DO REVERSE/DESCENDING ORDER  ADDING 'DESC'

```sql
SELECT TITLE, AUTHOR
FROM BOOKS
WHERE AUTHOR NOT LIKE '%ROWLING%'
ORDER BY AUTHOR DESC
```

WE CAN ORDER ON MULTIPLE COLUMNS

- First it will be ordered on the first column
- If multiple rows have same first column it will be internall reordered on second columns
- Here books will be first ordered by author name in descending order
- For books of same author they will be organized in the ascnending order of price

```sql
SELECT  TITLE, AUTHOR, PRICE 
FROM BOOKS
ORDER BY 
    AUTHOR DESC,
    PRICE 
```

## AGGREGATE QUERIES

  

- So far all queries handled each row independent of each other
    - They are either selected or filtered out
- But sometimes we may need to create result by processing rows with respect to each other
- We end up getting fewer rows
    - In many cases just 1.

  

### DISTINCT

- Returns unique values from a column

```sql
SELECT
    DISTINCT AUTHOR
FROM
    BOOKS
```

AGGREGATE RETURNING CALCULATION BASED ON COLUMNS

  

- COUNT
- SUM
- AVG
- MIN
- MAX

```sql
SELECT 
    COUNT(TITLE)
FROM
    BOOKS
```

# IMPORTANT NOTE

- Remember, All queries always returna  table of data
- TABLE IS ALWAYS RECTAGULAR
    - EACH ROW WILL HAVE SAME NUMBER OF COLUMNS
- WHEN WRITING OUR PROJECTION WE MUST REALISE THAT WE CANT SELECT THINGS THAT DON'T FORM A PERFECT REACTANGLE
    - EACH ROWS WILL HAVE ALL THE PROJECTED DATA

## WHAT WILL NOT WORK

### ATTEMPT TO SELECT DISTINCT AUTHOR AND TITLE

- DISTINCT AUTHOR WILL YIELD 18 ROWS
- TITLE WILL YIELD 43 ROWS
- DOESN'T FORM A PROPER RECTANGLE
- THINK, DOESN'T EVEN MAKE SENSE
- Here, distinct is ignored.

```sql
SELECT
    DISTINCT AUTHOR,
    TITLE
FROM
    BOOKS
```

### Again we can't use aggregate (returns 1 row) and non aggregate (returning mulitple rows)

```sql
SELECT
    DISTINCT AUTHOR,
    COUNT(AUTHOR)
FROM
    BOOKS
```

### WE CAN SAFELY COMBINE AGGREGATE RETURNING 1 ROWS

```sql
SELECT 
    COUNT(*),
    AVG(RATING),
    MIN(PRICE),
    MAX(PRICE),
    AVG(PRICE)
FROM 
    BOOKS
```

### PROBLEM: NO COLUMN NAME

# Aliasing

- We can provide an alias name for a column or a computed aggerate value
- NOTE:
    - we may use 'as' a filler word to make query look cleaner
    - It is opitonal and removing it only hurts the readability
- Aliasing is not just for clarity it can be used in our conditions or query also
- Aliasing can also be done for table name (as mentioned in from clause) or any other named entity

```sql
SELECT 
    COUNT(*) as TOTAL_BOOKS,
    AVG(RATING) AVERAGE_RATING,
    AVG(PRICE) AVERAGE_PRICE,
    MIN(PRICE) LOWEST_PRICE,
    MAX(PRICE) HIGHEST_PRICE
FROM 
    BOOKS
```
