## WE CAN USE AGGERAGATE FUNCTIONS

- They always returns a single row
- But when group by is applied aggeragate works on each group and not on the entire table
- So if we use
    - count(title)
        - It will not mean count all titles in the entire table
        - It means count title within each group

- Remeber when group by is used
    - SELECT APPLIES ON GROUP 
    - NOT ON THE MAIN TABLE

```sql
SELECT
    AUTHOR, /*GROUP KEY*/
    COUNT(TITLE) BOOKS_WRITTEN,  /*COUNTS TITLE WITHIN EACH GROUP*/
    AVG(RATING) AUTHOR_RATING /* AVG OF RATINGS FOR EACH ROW WITHIN THE GROUP*/
FROM
    BOOKS    /* PROJECTION DOESNT APPLY TO BOOKS TABLE*/
GROUP BY 
    AUTHOR  /* PROJECTIONA PPLIES TO EACH GROUP AND THE RESULT IS COLLECTED */
```

# WHEN GROUP BY IS ACTIVE

- <span style="font-size: 14px;">Most other clauses are also applied on the group table and not the real table</span>
    - ORDER BY
    - WHERE

```sql
SELECT 
    /* selected from group not from books*/
    AUTHOR,
    COUNT(TITLE) BOOKS_WRITTEN,
    AVG(RATING)  AVERAGE_RATING
FROM
    BOOKS
WHERE /* applies on original table before grouping starts*/
    TITLE NOT LIKE '%White Tiger%'
GROUP BY
    AUTHOR
    /* applies to group not on books */
HAVING /* similar to where but applies on the group*/
    COUNT(TITLE) >0
ORDER BY
    AVERAGE_RATING DESC
```
