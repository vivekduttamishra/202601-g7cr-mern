## WE DONT HAVE AUTHOR TABLE BUT WE CAN CONCEPTUALLY HAVE AUTHOR CENTRIC INFORMATION PULLED FROM CURRENT TABLE

\* Author Name

\* Books written

\* Average votes

\* Average review

```sql
SELECT
    AUTHOR AS NAME,
    COUNT(TITLE) AS BOOKS_WRITTEN,
    CAST(AVG(CAST(RATING AS FLOAT)) AS DECIMAL(4,1)) AS RATING,
    COUNT(RATING) AS VOTES
FROM
    BOOKS
LEFT JOIN
    REVIEWS
ON
    ID=BOOK_ID
GROUP BY
    AUTHOR
```

# <span style="font-size: 14px; color: var(--vscode-foreground);">PROBLEMS</span>

- A query is not persistent in database
    - you need to write it again and again
- Querying against a query is more complex
    - find all authors with votes greater than 10
        - WE MAY SOLVE THIS WITH HAVING
    - we may want to join some information from books and authors

# Solution: Views

1. The appear like a virtual table

- All queries are virtual table

3. They are persitent

- <span style="color: var(--vscode-foreground);">stored inside database for re-use</span>

5. They are dynamic
- when data changes in tables against which they are queries, they too change
7. <span style="color: var(--vscode-foreground);">They are like named queries&nbsp;</span>

```sql
CREATE VIEW AUTHORS
AS
-- Get a list of tables and views in the current database
SELECT
    AUTHOR AS NAME,
    COUNT(TITLE) AS BOOKS_WRITTEN,
    CAST(AVG(CAST(RATING AS FLOAT)) AS DECIMAL(4,1)) AS RATING,
    COUNT(RATING) AS VOTES
FROM
    BOOKS
LEFT JOIN
    REVIEWS
ON
    ID=BOOK_ID
GROUP BY
    AUTHOR
GO
```

## NOW WE CAN QUERY AGAINST A VIEW JUST AS WE DO AGAINST A TABLE

```sql
SELECT 
    NAME, RATING, VOTES
FROM
    AUTHORS
```

WE MAY FIND ALL TITLES BY AUTHORS SHOW RATING ARE MORE THAN 4 AND HAVE MORE THAN 10 VOTES

```sql
SELECT
    TITLE,
    AUTHOR,
    RATING,
    VOTES
FROM
    BOOKS B
JOIN
    AUTHORS A
ON
    B.AUTHOR=A.NAME
WHERE
    RATING>4 
    AND
    VOTES>10
```
