## BUT SINCE THEY ARE RELATED WE CAN TRY A COMBINED QUERY ON THEM

```sql
SELECT 
    TITLE,
    PRICE,
    RATING
FROM
    BOOKS, 
    REVIEWS
```

```sql
SELECT 
    BOOKS.TITLE TITLE,
    REVIEWS.TITLE REVIEWED_TITLE,
    PRICE,
    RATING
FROM
    BOOKS, 
    REVIEWS
```

## We can filter this large Table using Where

```sql
SELECT
    b.TITLE,
    PRICE,
    RATING
FROM
    BOOKS b,
    REVIEWS r
WHERE
    b.TITLE=r.TITLE
```

### WHERE CLAUSE TO COMBINE TABLES VIOLATES SINGLE RESPONSIBILITY

- It is supposed to filter records from a valid set of record
- But relational clause are ther eto comine the rows in a sensible way

# JOIN

- The official clause to join multiple tables

```sql
SELECT
    b.title,
    author,
    price,
    rating
FROM
    BOOKS b
JOIN
    REVIEWS r
ON
    b.title=r.title
WHERE
    author not like '%chetan bhagat%'
```

# JOIN TYPES

- There are few different types of propsed joins
    
- INNER JOIN
    
    - joins the table just like where clause did
    - it represents intersection of the two table
    - takes only if matching key exists on both tables
- LEFT (outer) join
    
- Takes all values from the left table (mentioned in from)
    
    - matches values from the right table (mentioned in join)
    - Any value that is present in left is considered
    - If matching value is not found in right, we use null for those cells
    - Any value that is not present in left is ignored
        - no rows is created
- RIGHT (outer) join
    
    - Just like left join
    - But includes values from right table
    - ignores values of the left table not having relation in the right
- Full Join
    
    - takes all values from both sides
    - fills null for not found
    - NOT SUPPORTED IN THE MAJORITY OF DATABASES

# Our Scenario

- Books table contain a title 'Brethren' but there is no corresponding review
- <span style="color: var(--vscode-foreground);">Reviews table contain reviews about a book titled 'Kurukshetra' but there is no book information in books table</span>

# INNER Join

- Neither of the two books will finda  mention in inner join
- Default join
- inner word is opitonal need not be included

```sql
SELECT 
    b.title,
    author,
    price,
    rating
FROM
    books b
INNER JOIN
    reviews r
ON
    b.title = r.Title
ORDER BY b.title
```

## 2\. Left Outer Join

- With left (outer will be implied)

- Will take all titles from books including 'brethren'
- But will ignore 'Kurukshetra' which is not present in left
- It makes a lot of sense to include books in our list even if it is not yet reviewed

```sql
SELECT 
    b.title,
    author,
    price,
    rating
FROM
    BOOKS B
LEFT JOIN
    REVIEWS r
ON
    b.title=r.title

ORDER BY TITLE
```

## 2\. Right Outer Join

- With right (outer will be implied)

- Will skip the  titles from books Like 'brethren' if not present on the right
- But take  'Kurukshetra' even if  not present in left
- Not too much of use
- <span style="font-size: 14px;">Left alone was sufficient</span>
    - <span style="font-size: 14px;">we could always swap the order to get the same result</span>
- In real world we may not need reivews of a book that doesn't exist

```sql
SELECT 
    b.title,
    author,
    price,
    r.title,
    rating
FROM
    BOOKS B
RIGHT JOIN
    REVIEWS r
ON
    b.title=r.title

ORDER BY b.TITLE
```
