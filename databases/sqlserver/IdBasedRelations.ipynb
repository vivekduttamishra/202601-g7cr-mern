# JOIN

- The official clause to join multiple tables

```sql
SELECT
    b.title,
    author,
    price,
    rating
FROM
    BOOKS b
JOIN
    REVIEWS r
ON
    id=book_id
WHERE
    author not like '%chetan bhagat%'
```

# JOIN TYPES

- There are few different types of propsed joins
    
- INNER JOIN
    
    - joins the table just like where clause did
    - it represents intersection of the two table
    - takes only if matching key exists on both tables
- LEFT (outer) join
    
- Takes all values from the left table (mentioned in from)
    
    - matches values from the right table (mentioned in join)
    - Any value that is present in left is considered
    - If matching value is not found in right, we use null for those cells
    - Any value that is not present in left is ignored
        - no rows is created
- RIGHT (outer) join
    
    - Just like left join
    - But includes values from right table
    - ignores values of the left table not having relation in the right
- Full Join
    
    - takes all values from both sides
    - fills null for not found
    - NOT SUPPORTED IN THE MAJORITY OF DATABASES

# Our Scenario

- Books table contain a title 'Brethren' but there is no corresponding review

# INNER Join

- Neither of the two books will finda  mention in inner join
- Default join
- inner word is opitonal need not be included

```sql
SELECT 
    b.title,
    author,
    price,
    rating
FROM
    books b
INNER JOIN
    reviews r
ON
    id=book_id
ORDER BY b.title
```

## 2\. Left Outer Join

- With left (outer will be implied)

- Will take all titles from books including 'brethren'
- But will ignore 'Kurukshetra' which is not present in left
- It makes a lot of sense to include books in our list even if it is not yet reviewed

```sql
SELECT 
    b.title,
    author,
    price,
    rating
FROM
    BOOKS B
LEFT JOIN
    REVIEWS r
ON
   id=book_id

ORDER BY TITLE
```

## 2\. Right Outer Join

- With right (outer will be implied)

- Will skip the  titles from books Like 'brethren' if not present on the right
- But take  'Kurukshetra' even if  not present in left
- Not too much of use
- <span style="font-size: 14px;">Left alone was sufficient</span>
    - <span style="font-size: 14px;">we could always swap the order to get the same result</span>
- In real world we may not need reivews of a book that doesn't exist

```sql
SELECT 
    b.title,
    author,
    price,
    book_id
    rating
FROM
    BOOKS B
RIGHT JOIN
    REVIEWS r
ON
    id=book_id

ORDER BY b.TITLE
```

# FOREIGN KEY CONSTRAINT

- FOREIGN KEY CONSTRAING ACTUALLY FORMS A PARENT CHILD RELATIONSHIP
- A CHILD TABLE CONTAINS ADDITIONAL INFORMATION RELATED TO INFORMATION PRESENT IN ANOTHER TABLE
- EXAMPLE: REVIEWS ARE ASSOCIATED WITH BOOK
    - BOOK IS PARENT
        - REVIEW IS CHILD

- FOREIGN KEY MAPS A COLUM  OF THE CHILD TO PRIMARY KEY OF THE PARENT

## FOREIGN KEY ADVANTAGE

- FK must map to a valid exisitng primary key of the parent at all times
- We can't add a new record that doesn't map to an exisitng aprent record
- We can't remove the primary key from parent  without actually removing all child rows from child table

  

## LIKE ALL CONSTRAINTS YOU CANT APPLY THIS CONSTRAINT UNTIL IT IS ACCEPTABLE TO ALL THE CURRENT ROWS.

```sql
ALTER TABLE REVIEWS
    ADD CONSTRAINT REVIEWS_FK_BOOK_ID FOREIGN KEY(BOOK_ID) REFERENCES BOOKS(ID)
```

## ALTERNATIVE SOLUTION:   CASCADING DELETE AND UPDATE

  

- WE CAN SET CASCADE\_DELETE AND UPDATE ON A FK CONSTRAINT
- THIS ENSURE 
    - CASCADE\_DELETE
        - IF PRIMARY ROW IS DELETED SECONDARY TABLE ROWS ARE ALSO DELETED
        - IF BOOK IS GONE, REVIEWS ARE MEANINGLESS
    - CASECADE\_UPDATE
        - IF PK IS UPDATED IN PRIMARY TABLE ALL REFERENCE IN CHILD TABLE IS UPDATED
            - CHANGE ID tiggers changes to BOOK\_ID

### BUT WE ALREADY HAVE CONSTRAINT

- WE HAVE TO REMOVE AND RECREATE

```sql
ALTER TABLE REVIEWS
    DROP CONSTRAINT REVIEWS_FK_BOOK_ID_2
```

```sql
ALTER TABLE REVIEWS
    ADD CONSTRAINT REVIEWS_FK_BOOK_ID
    FOREIGN KEY(BOOK_ID)
    REFERENCES BOOKS(ID)
    ON UPDATE CASCADE
    ON DELETE CASCADE
    
```
